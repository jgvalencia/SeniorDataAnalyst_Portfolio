### Import Pandas Seaborn and Matplotlib

### Correlation

```python
# df.corr() # ValueError: could not convert string to float: 'AFG'

### Correlation
df.corr(numeric_only=True)
```

### Heatmap selecting only numeric to avoid errors

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.rcParams['figure.figsize'] = (20, 8)
plt.show()
```

### df.groupby("Continent").mean() # TypeError: dtype 'str' does not support operation 'mean'

```python

df.groupby("Continent").mean(numeric_only=True)
```

### Select only one Continent Oceania

```python
df[df['Continent'].str.contains('Oceania')]
```

### Groupby only numeric values sorting by "2022 Population" with ascending=false

```python
df.groupby("Continent").mean(numeric_only=True).sort_values(by="2022 Population", ascending=False)
```

### Save DF to DF2 to generate some graphics

```python
df2 = df.groupby("Continent")[['1970 Population',
       '1980 Population', '1990 Population', '2000 Population',
       '2010 Population', '2015 Population', '2020 Population',
       '2022 Population']].mean(numeric_only=True).sort_values(by="2022 Population", ascending=False)
df2
```

```python
df.columns
```

### We will use PLOT to visualize some data

```python
df2.plot()
```

### Transpose Column to rows and create a new dataset

```python
df3 = df2.transpose()
df3
```

### Plot with Df3 dataset Asia Population has been growth since 1980 more than any othe continent

```python
df3.plot()
```
